```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/seasonal_agriculture_performance_dataset.csv")
```

```python
# checking missing values
df.isnull().sum()
```

```python
# checking the duplicated values
df.duplicated().sum()
```

```python
# fixing the missing values
df["Rainfall_mm"]=df["Rainfall_mm"].fillna(df["Rainfall_mm"].median())
df["Soil_Moisture_pct"]=df["Soil_Moisture_pct"].fillna(df["Soil_Moisture_pct"].median())
df["Yield_Tonnes_Ha"]=df["Yield_Tonnes_Ha"].fillna(df["Yield_Tonnes_Ha"].median())
```

```python
# Which season has the highest average crop yield?
season_yield=df.groupby("Season")["Yield_Tonnes_Ha"].mean()
season_yield
```

```python
season_yield.plot(kind="bar")
plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield(Tonnes/Ha)")
plt.show()
```

**Insight:** Kharif season records the highest average crop yield, followed by Rabi, while Zaid has the lowest average yield among the three seasons.

```python
# Which season is the most profitable?
season_profit=df.groupby("Season")["Profit_INR"].mean()
season_profit
```

```python
season_profit.plot(kind="bar")
plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.show()
```

**Insight:** Kharif season has the highest average profit, while Rabi has a considerably lower average profit. Zaid season shows a negative average profit, indicating that the average costs exceeded the average revenue during this season.

```python
# How does total agricultural production vary across seasons?
season_production=df.groupby("Season")["Production_Tonnes"].sum()
season_production
```

```python
season_production.plot(kind="bar")
plt.title("Total agricultural Production by Season")
plt.xlabel("Season")
plt.ylabel("Total Production (Tonnes)")
plt.xticks(rotation=0)
plt.show()
```

**Insight:**Kharif season has the highest total agricultural production, followed by Rabi, while Zaid has the lowest production. Kharif is leading in both average yield and total production. It also has the highest average profit.

```python
# Which season uses water most efficiently?
season_water_efficiency=df.groupby("Season")["Water_Efficiency_t_per_1000m3"].mean()
season_water_efficiency
```

```python
season_water_efficiency.plot(kind="bar")
plt.title("Average Water Efficiency by Season")
plt.xlabel("Season")
plt.ylabel("Water Efficiency (Tonnes per 1000 m^3)")
plt.xticks(rotation=0)
plt.show()
```

**Insigth:** Kharif shows the highest average water-use efficiency, while Zaid has the lowest. This indicates that agriculture output per unit of water varies across seasons.with Kharif making more efficient use of available water.

```python
# Does rainfall affect crop yield?
rainfall_yield_corr=df["Rainfall_mm"].corr(df["Yield_Tonnes_Ha"])
rainfall_yield_corr
```

```python
plt.scatter(df["Rainfall_mm"],df["Yield_Tonnes_Ha"])
plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Crop Yield (Tonnes/Ha)")
plt.show()
```

**Insight:**The correlation between rainfall and crop yield is approximately 0.031, indicates a very weak posotive linear relationship. Therefore, rainfall alone does not apper to be a strong predictor of crop yield in this dataset.

```python
# Temperature vs Crop Yield
temperature_yield_corr=df["Avg_Temperature_C"].corr(df["Yield_Tonnes_Ha"])
temperature_yield_corr
```

```python
plt.scatter(df["Avg_Temperature_C"], df["Yield_Tonnes_Ha"])
plt.title("Temperature vs Crop Yield")
plt.xlabel("Temperature (C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

**Insight:** The correlation between average temperature and crop yield is approximately 0.01, indicating an extremely weak positive linear relationship. This suggests that temperature alone does not have a meaningful linear association with crop yield in the dataset.

```python
# Which crop perform best in each season?
crop_season_yield=df.groupby(["Season", "Crop"])["Yield_Tonnes_Ha"].mean()
crop_season_yield
```

```python
crop_season_yield.unstack().plot(kind='bar')

plt.title('Average Crop Yield by Crop and Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=0)
plt.legend(title='Crop')
plt.show()
```

**Insight:** Sugarcane has the highest average yield across all three seasons. Its average yield is highest in Kharif (53.46 tonnes/ha), followed by Rabi (43.29 tonnes/ha) and Zaid (38.42 tonnes/ha), indicating a decline in Sugarcane productivity across seasons.

```python
# Cost, Revenue and Profit
season_economics = df.groupby('Season')[
    ['Total_Cost_INR', 'Revenue_INR', 'Profit_INR']
].mean()

season_economics
```

```python
season_economics.plot(kind='bar')

plt.title('Average Cost, Revenue and Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Amount (INR)')
plt.xticks(rotation=0)
plt.legend(title='Economic Measure')
plt.show()
```

**Insight:** Kharif has the highest average revenue and profit among the three seasons. Rabi remains profitable but generates substantially lower returns, while Zaid records a negative average profit because its average revenue is lower than its average cost.
